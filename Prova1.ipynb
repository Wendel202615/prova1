```python
idade = int(input("Qual sua idade? "))
valor_total_carteira = float(input('Qual valor total na carteira? '))
if idade >= 16:
  print('entrada permitida')
else:
  print('entrada somente com acompanhante')
if valor_total_carteira >= 30:
  print('ingresso comprado, bom filme!')
else:
  print('saldo insuficiente para o ingresso')
print('Foi bom te atender')
```
